```python
mlb = mlbstatsapi.Mlb()
```

```python
name = "Shohei Ohtani"
player = mlb.get_people_id(name)[0]
```

```python
stats = ["season"]
groups = ["hitting"]
params = {"seasonAdvanced": 2025}
stats = mlb.get_player_stats(player, stats, groups, **params)
```

```python
c_woba_2025 = {
    "ubb": 0.691,
    "hbp": 0.722,
    "1b": 0.882,
    "2b": 1.252,
    "3b": 1.584,
    "hr": 2.037,
    "runsb": 0.2,
    ".runcs": -0.41,
    "rppa": 0.118,
    "rpw": 9.774,
    "cfip": 3.135,
}
```

```python
simple_stats = stats["hitting"]["season"].splits[0].stat
```

```python
c_woba = c_woba_2025
ubb = simple_stats.base_on_balls - simple_stats.intentional_walks
singles = simple_stats.hits - simple_stats.doubles - simple_stats.triples - simple_stats.home_runs

woba = (
    c_woba["ubb"] * ubb
    + c_woba["hbp"] * simple_stats.hit_by_pitch
    + c_woba["1b"] * singles
    + c_woba["2b"] * simple_stats.doubles
    + c_woba["3b"] * simple_stats.triples
    + c_woba["hr"] * simple_stats.home_runs
) / (
    simple_stats.at_bats
    + simple_stats.base_on_balls
    - simple_stats.intentional_walks
    + simple_stats.sac_flies
    - simple_stats.hit_by_pitch
)
woba
```
